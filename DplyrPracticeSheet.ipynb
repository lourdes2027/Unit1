```r
mpg_modified <- mpg %>%
  filter(year==1999 & manufacturer=="audi") %>%
  select(manufacturer, year, cty, hwy) %>%
  mutate(cty_feet = cty*5280, hwy_feet = hwy*5280)

mpg_modified

```

Above, I am creating a new dataset called `mpg_modified`, which includes only cars manufactured by audi in 1999, and only displays info on manufacturer, year, city miles per gallon, and highway miles per gallon, with an additional two variables which are the prior two converted to feet per gallon.
